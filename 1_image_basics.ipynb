```python
import cv2 as cv
import matplotlib .pyplot as plt

im = cv.imread("images/rgbPowder.jpeg")
assert im is not None

im_blue= im.copy()
im_blue[:,:,1]=0
im_blue[:,:,2]=0

im_green= im.copy()
im_green[:,:,0]=0
im_green[:,:,2]=0

im_red= im.copy()
im_red[:,:,0]=0
im_red[:,:,1]=0

fig, ax =plt.subplots(2,2)
ax[0,0].set_title('original')
ax[0,0].imshow(cv.cvtColor(im, cv.COLOR_BGR2RGB))
ax[0,1].imshow(cv.cvtColor(im_blue, cv.COLOR_BGR2RGB))
ax[1,0].imshow(cv.cvtColor(im_green, cv.COLOR_BGR2RGB))
ax[1,1].imshow(cv.cvtColor(im_red, cv.COLOR_BGR2RGB))

for i in range(2):
    for j in range(2):
        ax[i,j].axis('off')

plt.show()
```

```python
import cv2 as cv
import numpy as np
import matplotlib .pyplot as plt

im1 = cv.imread("images/Albert Einstein.jpeg" ,cv.IMREAD_GRAYSCALE)
assert im1 is not None

t= np.array([255-i for i in range(256)], dtype=np.uint8) # defining the tranformation
#im2=t[im1]
im2= cv.LUT(im1,t) # above and cv.LUT both are the same

im3=  cv.equalizeHist(im1)

fig, ax =plt.subplots(1,3, figsize=(12,4))
ax[0].imshow(im1, cmap='grey')
ax[0].set_title("original")
ax[0].axis('off')


ax[1].plot(np.arange(256),t,color='black')
ax[1].set_title("brightened")

ax[2].imshow(im2, cmap='grey')
ax[2].set_title("transformed")
ax[2].axis('off')

plt.show()
```

```python
import cv2 as cv
import numpy as np
import matplotlib .pyplot as plt

im1 = cv.imread("images/Albert Einstein.jpeg" ,cv.IMREAD_GRAYSCALE)
assert im1 is not None

im2=  cv.equalizeHist(im1)

fig, ax =plt.subplots(1,2, figsize=(12,6))
ax[0].imshow(im1, cmap='grey')
ax[0].set_title("original")
ax[0].axis('off')

ax[1].imshow(im2, cmap='grey')
ax[1].set_title("Equal Histogram")
ax[1].axis('off')

plt.show()
```
