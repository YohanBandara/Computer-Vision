# Averaging Kernel

```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt

#im=cv.imread("Images/Sigiriya.jpeg")
im = cv.imread('images/sigiriya.jpeg', cv.IMREAD_GRAYSCALE)
assert im is not None

kernel= np.ones((3,3), dtype=np.float32)/9

imavg= cv.filter2D(im,cv.CV_32F,kernel)

fig, ax= plt.subplots(1,2)
ax[0].imshow(im, cmap='gray', vmin=0, vmax=255)
ax[0].set_title("original_image")
ax[1].imshow(imavg, cmap='gray', vmin=0, vmax=255)
ax[1].set_title("averaged_image")

for a in ax:
    a.axis('off')

plt.show()


```

```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt


im = cv.imread('images/sigiriya.jpeg', cv.IMREAD_GRAYSCALE)
assert im is not None

#kernel= np.ones((3,3), dtype=np.float32)/9
sobel_x= np.array([[-1,-2,-1],[0,0,0],[1,2,1]],dtype=np.float32)
sobel_y= np.array([[-1,0,1],[-2,0,2],[-1,0,1]],dtype=np.float32)
print(sobel_x)
print(sobel_y)

#imavg= cv.filter2D(im,cv.CV_32F,kernel)
im_sobel_x= cv.filter2D(im,cv.CV_32F,sobel_x)
im_sobel_y= cv.filter2D(im,cv.CV_32F,sobel_y)

#im_sobel_x= cv.normalize(im_sobel_x, None, 0, 255, cv.NORM_MINMAX, cv.CV_8U)
#im_sobel_y= cv.normalize(im_sobel_y, None, 0, 255, cv.NORM_MINMAX, cv.CV_8U)

fig, ax= plt.subplots(1,3, figsize=(14,10))
ax[0].imshow(im, cmap='gray', vmin=0, vmax=255)
ax[0].set_title("original_image")
ax[1].imshow(im_sobel_x, cmap='gray')
ax[1].set_title("Sobel X")
ax[2].imshow(im_sobel_y, cmap='gray')
ax[2].set_title("Sobel Y")

for a in ax:
    a.axis('off')

plt.show()
```

```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt

#im=cv.imread("Images/Sigiriya.jpeg")
im = cv.imread('images/tom.jpg', cv.IMREAD_GRAYSCALE)
assert im is not None

blurred= cv.GaussianBlur(im, (21,21),0)       #
high_pass=cv.subtract(im,blurred)
sharpened= cv.add(im,high_pass)               #

fig, ax= plt.subplots(1,4, figsize=(15,5))
ax[0].imshow(im, cmap='gray')
ax[0].set_title("original_image")
ax[1].imshow(blurred, cmap='gray')
ax[1].set_title("blurred_image")
ax[2].imshow(high_pass, cmap='gray')
ax[2].set_title("high_pass")
ax[3].imshow(sharpened, cmap='gray')
ax[3].set_title("sharpened")

for a in ax:
    a.axis('off')

plt.show()
```

# Creaing Gaussian Filter Kerenel

```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

sigma =2.0
x= np.arange(-5,5.1,0.1)
y= np.arange(-5,5.1,0.1)
X,Y =np.meshgrid(x,y)
print(X)
print(Y)

g=1/(2*np.pi*sigma**2)*np.exp(-(X**2+Y**2)/(2*sigma**2))

fig =plt.figure(figsize=(10,10))
ax = fig.add_subplot(111, projection='3d')
ax.plot_surface(X,Y,g,cmap='viridis')

plt.axis('off')
plt.show()


```
